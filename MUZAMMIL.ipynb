```python
class BankAccount:
    bank_name = "AI Bank"                 # class attribute (shared)

    def __init__(self, owner, balance=0):
        self.owner = owner                # instance attributes
        self._balance = balance           # _ means "internal, don't touch directly"

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount

    def withdraw(self, amount):
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount

    def __str__(self):
        return f"{self.owner}: Rs {self._balance:,}"

    def __repr__(self):
        return f"BankAccount(owner={self.owner!r}, balance={self._balance})"

acc = BankAccount("Sadia", 1000)
acc.deposit(500)
acc.withdraw(200)
print(acc)
print(repr(acc))
print(acc.balance, BankAccount.bank_name)
```

```python

class Student:

    school_name = "SSUET "

    def __init__(self, name, roll_no, marks=0):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def add_marks(self, amount):
        self.marks += amount
        print(amount, "marks added to", self.name)

    def deduct_marks(self, amount):
        self.marks -= amount
        print(amount, "marks deducted from", self.name)

    def display_info(self):
        print("\nStudent Name:", self.name)
        print("Roll Number:", self.roll_no)
        print("Marks:", self.marks)
        print("School Name:", Student.school_name)

    def check_result(self):
        if self.marks >= 50:
            print("Result: Pass")
        else:
            print("Result: Fail")

student1 = Student("Bilal Mahmood Ansari", 297, 70)
student2 = Student("Muzammil", 301, 85)
student3 = Student("Wahaj", 306, 60)
student4 = Student("Ahmed", 309, 70)

student1.add_marks(20)
student2.add_marks(10)
student3.add_marks(15)
student4.deduct_marks(25)


student1.display_info()
student2.display_info()
student3.display_info()
student4.display_info()

print("----- Student Results -----")

student1.check_result()
student2.check_result()
student3.check_result()
student4.check_result()
```
